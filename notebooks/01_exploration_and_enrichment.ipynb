```python
import sys
import os
import pandas as pd

# Add project root to path to import from src
sys.path.append(os.path.abspath(os.path.join('..')))

from src.data_loader import load_unified_data

# Define path and load
data_path = "../data/raw/ethiopia_fi_unified_data.xlsx"
df = load_unified_data(data_path)

# Quick look at the structure (Record Types and Indicators)
print("\n--- Record Type Counts ---")
print(df['record_type'].value_counts())

# Identify Gaps (Gap Analysis)
observations = df[df['record_type'] == 'observation'].copy()
observations['year'] = pd.to_datetime(observations['observation_date']).dt.year

gap_matrix = observations.pivot_table(
    index='indicator_code', 
    columns='year', 
    values='value_numeric', 
    aggfunc='count'
).fillna(0)

print("\n--- Data Gap Matrix ---")
display(gap_matrix)
```

```python
# Create enrichment records as a list of dictionaries
enrichment_data = [
    {'record_type': 'observation', 'indicator_code': 'USG_TELEBIRR_USERS', 'value_numeric': 27.2, 'observation_date': '2022-12-31', 'source_name': 'Ethio Telecom'},
    {'record_type': 'observation', 'indicator_code': 'USG_TELEBIRR_USERS', 'value_numeric': 41.0, 'observation_date': '2023-12-31', 'source_name': 'Ethio Telecom'},
    {'record_type': 'observation', 'indicator_code': 'USG_MPESA_USERS', 'value_numeric': 3.1, 'observation_date': '2023-12-31', 'source_name': 'Safaricom Ethiopia'}
]

# Convert to DataFrame and append
enrichment_df = pd.DataFrame(enrichment_data)
df_enriched = pd.concat([df, enrichment_df], ignore_index=True)

# Verify the addition
print(f"Total records after enrichment: {len(df_enriched)}")
print(df_enriched.tail(3))
```
